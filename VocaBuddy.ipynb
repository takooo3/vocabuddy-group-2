```python

import random

vocab_list = {
    "yyc": "歪歪西",
    "wpc": "奇異果",
    "tako": "張語珊",
    "juanjuan": "水豚"
}

def get_random_word():
    random_word = random.choice(list(vocab_list.keys()))
    definition = vocab_list[random_word]
    print("Luca thinks the below word is cute. ")
    print(f"單字：{random_word}")
    print(f"解釋：{definition}")

get_random_word()
```
